```python
import os
from collections import Counter

IMAGE_DIR = "../detection_dataset/images/train"
LABEL_DIR = "../detection_dataset/labels/train"

IMAGE_EXTENSIONS = (".jpg", ".jpeg", ".png")

missing_labels = []
missing_images = []
empty_labels = []
invalid_labels = []

class_counter = Counter()

total_images = 0
total_annotations = 0
```

```python
image_files = [
    f for f in os.listdir(IMAGE_DIR)
    if f.lower().endswith(IMAGE_EXTENSIONS)
]

label_files = [
    f for f in os.listdir(LABEL_DIR)
    if f.endswith(".txt")
]

print(f"Total Image Files: {len(image_files)}")
print(f"Total Label Files: {len(label_files)}")
```

```python
for image_file in image_files:

    total_images += 1

    image_name = os.path.splitext(image_file)[0]
    label_file = image_name + ".txt"

    label_path = os.path.join(LABEL_DIR, label_file)

    if not os.path.exists(label_path):
        missing_labels.append(image_file)
        continue

    if os.path.getsize(label_path) == 0:
        empty_labels.append(label_file)
        continue

    with open(label_path, "r") as f:
        lines = f.readlines()

    for line in lines:

        parts = line.strip().split()

        if len(parts) != 5:
            invalid_labels.append(label_file)
            continue

        class_id = parts[0]

        class_counter[class_id] += 1
        total_annotations += 1
```

```python
import os
from collections import Counter

# =====================================
# DATASET PATHS
# =====================================

IMAGE_DIRS = [
    "../detection_dataset/images/train",
    "../detection_dataset/images/val",
    "../detection_dataset/images/test"
]

LABEL_DIRS = [
    "../detection_dataset/labels/train",
    "../detection_dataset/labels/val",
    "../detection_dataset/labels/test"
]

IMAGE_EXTENSIONS = (".jpg", ".jpeg", ".png")

# =====================================
# COUNTERS
# =====================================

missing_labels = []
missing_images = []
empty_labels = []
invalid_labels = []

class_counter = Counter()

total_images = 0
total_annotations = 0

# =====================================
# CHECK ALL SPLITS
# =====================================

for IMAGE_DIR, LABEL_DIR in zip(IMAGE_DIRS, LABEL_DIRS):

    image_files = [
        f for f in os.listdir(IMAGE_DIR)
        if f.lower().endswith(IMAGE_EXTENSIONS)
    ]

    label_files = [
        f for f in os.listdir(LABEL_DIR)
        if f.endswith(".txt")
    ]

    # -----------------------------
    # IMAGE -> LABEL
    # -----------------------------

    for image_file in image_files:

        total_images += 1

        image_name = os.path.splitext(image_file)[0]
        label_file = image_name + ".txt"

        label_path = os.path.join(LABEL_DIR, label_file)

        # Missing label
        if not os.path.exists(label_path):
            missing_labels.append(image_file)
            continue

        # Empty label
        if os.path.getsize(label_path) == 0:
            empty_labels.append(label_file)
            continue

        # Read labels
        with open(label_path, "r") as f:
            lines = f.readlines()

        for line in lines:

            parts = line.strip().split()

            # YOLO format check
            if len(parts) != 5:
                invalid_labels.append(label_file)
                continue

            class_id = parts[0]

            class_counter[class_id] += 1
            total_annotations += 1

    # -----------------------------
    # LABEL -> IMAGE
    # -----------------------------

    for label_file in label_files:

        label_name = os.path.splitext(label_file)[0]

        image_found = False

        for ext in IMAGE_EXTENSIONS:

            image_path = os.path.join(
                IMAGE_DIR,
                label_name + ext
            )

            if os.path.exists(image_path):
                image_found = True
                break

        if not image_found:
            missing_images.append(label_file)

# =====================================
# FINAL REPORT
# =====================================

print("\n========== FULL DATASET REPORT ==========")

print(f"\nTotal Images: {total_images}")
print(f"Total Annotations: {total_annotations}")

print("\nClass Distribution:")
print(f"Healthy (0): {class_counter['0']}")
print(f"Unhealthy (1): {class_counter['1']}")

print("\nMissing Labels:", len(missing_labels))
print("Empty Labels:", len(empty_labels))
print("Invalid Labels:", len(invalid_labels))
print("Missing Images:", len(missing_images))

# =====================================
# OPTIONAL DETAILS
# =====================================

if missing_labels:

    print("\n========== MISSING LABEL FILES ==========\n")

    for item in missing_labels:
        print(item)
```

```python
print("\n========== MISSING IMAGE FILES ==========\n")

if len(missing_images) == 0:
    print("No missing images found!")

else:
    for i, item in enumerate(missing_images, start=1):
        print(f"{i}. {item}")
```

```python
from ultralytics import YOLO
import pandas as pd
import matplotlib.pyplot as plt

# LOAD MODEL
model = YOLO(
    "../runs/detect/models/trained/bitol_yolov8/weights/best.pt"
)

print("Model loaded successfully!")

# TEST EVALUATION
metrics = model.val(
    data="../detection_dataset/data.yaml",
    split="test"
)

print("Test evaluation completed!")

# OVERALL RESULTS
map50 = metrics.box.map50
map5095 = metrics.box.map
precision = metrics.box.mp
recall = metrics.box.mr

results_df = pd.DataFrame({
    "Metric": [
        "Precision",
        "Recall",
        "mAP50",
        "mAP50-95"
    ],
    "Score": [
        precision,
        recall,
        map50,
        map5095
    ]
})

display(results_df)

# CLASS-WISE RESULTS
class_names = ["healthy", "unhealthy"]

class_df = pd.DataFrame({
    "Class": class_names,
    "Precision": metrics.box.p,
    "Recall": metrics.box.r,
    "mAP50": metrics.box.ap50,
    "mAP50-95": metrics.box.ap
})

display(class_df)

# PLOT
plt.figure(figsize=(8,5))
plt.bar(results_df["Metric"], results_df["Score"])
plt.ylim(0,1)
plt.title("Overall Test Performance")
plt.show()

# PREDICTION
results = model.predict(
    source="../detection_dataset/images/test",
    save=True,
    conf=0.5
)

print("Prediction completed!")
```
